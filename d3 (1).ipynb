```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\manoj\Downloads\zomato.csv\zomato.csv")
```

```python
df
```

```python

df.drop(columns="address", inplace=True)
df.drop(columns="phone", inplace=True)
df.drop(columns="dish_liked", inplace=True)
```

```python
df.drop(columns="reviews_list", inplace=True)
```

```python
df.drop(columns="menu_item", inplace=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["rate"]=df["rate"].str.replace('/5',' ')
```

```python
df.fillna(df.mode().iloc()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.rename(columns={"listed_in(type)":"type"}, inplace=True)
```

```python
df.rename(columns={"listed_in(city)":"city"}, inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df["rate"]=pd.to_numeric(df["rate"], errors="coerce")
```

```python
df.info()
```

```python
df["approx_cost(for two people)"]=pd.to_numeric(df["approx_cost(for two people)"], errors="coerce")
```

```python
df.rename(columns={"approx_cost(for two people)":"approx_cost"}, inplace=True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
plt.hist(df["votes"])
plt.title("Votes Distribution")
plt.xlabel("Votes")
plt.ylabel("Frequency")
plt.show()
```

```python
top10 = df.sort_values("votes", ascending=False).head(10)

plt.figure(figsize=(10,6))

plt.bar(top10["name"], top10["votes"])

plt.xticks(rotation=90)

plt.title("Top 10 Restaurants by Votes")
plt.xlabel("Restaurant Name")
plt.ylabel("Votes")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="online_order",
            y="votes",
            data=df)

plt.title("Online Order vs Votes")

plt.show()
```
